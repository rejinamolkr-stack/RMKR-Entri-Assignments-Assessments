```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = sns.load_dataset("titanic")
```

```python
plt.figure(figsize=(9,6))
sns.countplot(data=df, x='embarked', hue='sex', palette='Set2')

plt.title("Passengers Boarding at Each Port")
plt.xlabel("Embarked Port")
plt.ylabel("Passenger Count")
plt.legend(title="Gender")
plt.show()

print("Observation:")
print("- Most passengers boarded from Southampton (S).")
print("- Male passengers are more than female passengers at every port.")
print("- Queenstown (Q) has the lowest number of passengers.\n")
```

```python
plt.figure(figsize=(10,6))
sns.swarmplot(data=df, x='pclass', y='age', hue='sex', palette='Dark2')

plt.title("Age Distribution Across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.legend(title="Gender")
plt.show()

print("Observation:")
print("- First-class passengers generally include older people.")
print("- Third class has many younger passengers.")
print("- Both males and females are present in all passenger classes.\n")
```

```python
gender = df['sex'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    gender,
    labels=gender.index,
    autopct='%1.1f%%',
    explode=(0,0.1),
    colors=['skyblue','lightcoral'],
    startangle=120,
    shadow=True
)

plt.title("Gender Distribution of Titanic Passengers")
plt.show()

print("Observation:")
print("- Male passengers make up the majority of the dataset.")
print("- Female passengers represent a smaller percentage.")
print("- The exploded slice highlights the female passenger group.\n")
```

```python
correlation = df[['age','fare','pclass','sibsp','parch','survived']].corr()

plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True, cmap='YlGnBu', fmt=".2f")

plt.title("Correlation Matrix")
plt.show()

print("Observation:")
print("- Passenger class and fare have a strong negative correlation.")
print("- Higher fares are slightly associated with survival.")
print("- Age has only weak correlations with other features.\n")
```

```python
plt.figure(figsize=(10,6))
sns.violinplot(
    data=df,
    x='pclass',
    y='fare',
    hue='sex',
    split=True,
    palette='Set3'
)

plt.title("Fare Distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.legend(title="Gender")
plt.show()

print("Observation:")
print("- First-class passengers paid the highest fares.")
print("- Third-class passengers mostly paid lower fares.")
print("- Fare distributions differ slightly between males and females.\n")
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x='age',
    y='fare',
    hue='survived',
    style='survived',
    s=70
)

plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")
plt.show()

print("Observation:")
print("- Passengers who paid higher fares had better survival chances.")
print("- Most passengers paid lower fares.")
print("- Age does not show a strong relationship with fare.")
```
