```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

```

```python
#Load Dataset & Impute Missing Data
data_url = "winequality-red.csv" 

try:
    df = pd.read_csv(data_url, sep=';')
except Exception:
    df = pd.read_csv(data_url)

print("Data Preprocessing")
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")

missing_before = df.isnull().sum().sum()
print(f"Total missing values before imputation: {missing_before}")

df.fillna(df.mean(), inplace=True)

missing_after = df.isnull().sum().sum()
print(f"Total missing values after mean imputation: {missing_after}\n")

```

```python
#Extract Columns as Vectors
print("Vector Extraction")
alcohol_vector = df['alcohol'].to_numpy()
citric_acid_vector = df['citric acid'].to_numpy()

print(f"Alcohol Vector Shape: {alcohol_vector.shape}")
print(f"First 5 elements of Alcohol vector: {alcohol_vector[:5]}")
print(f"Citric Acid Vector Shape: {citric_acid_vector.shape}")
print(f"First 5 elements of Citric Acid vector: {citric_acid_vector[:5]}\n")

```

```python
#Calculate Covariance Matrix
print("Covariance Matrix")
selected_features = ['alcohol', 'density']
X = df[selected_features].values  # Shape: (1599, 2)

cov_matrix = np.cov(X.T)

print("Covariance Matrix [Alcohol, Density]:")
print(cov_matrix)
print(f"Variance (Alcohol): {cov_matrix[0, 0]:.6f}")
print(f"Variance (Density): {cov_matrix[1, 1]:.6f}")
print(f"Covariance (Alcohol, Density): {cov_matrix[0, 1]:.6f}\n")

```

```python
# ==========================================
# Question 4: Eigen Decomposition
# ==========================================
print("--- Question 4: Eigen Decomposition ---")
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Sort eigenvalues and eigenvectors in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]
top_eigenvalues = eigenvalues[sorted_indices]
top_eigenvectors = eigenvectors[:, sorted_indices]

print("Top 2 Eigenvalues:")
for i, val in enumerate(top_eigenvalues):
    print(f"  Eigenvalue {i+1}: {val:.6f}")

print("\nCorresponding Eigenvectors (as columns):")
print(top_eigenvectors)
print(f"  v1 (Direction of Max Variance): {top_eigenvectors[:, 0]}")
print(f"  v2 (Orthogonal Direction): {top_eigenvectors[:, 1]}\n")

```

```python
# ==========================================
# Question 5: Distribution of Wine Quality
# ==========================================
print("--- Question 5: Wine Quality Distribution ---")
quality_distribution = df['quality'].value_counts().sort_index()
most_common_quality = df['quality'].mode()[0]

print("Wine Quality Frequency Table:")
print(quality_distribution)
print(f"\nMost Common Wine Quality Score: {most_common_quality}\n")

```
