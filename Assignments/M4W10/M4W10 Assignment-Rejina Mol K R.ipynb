
**Dataset:**
Breast Cancer dataset available in `sklearn`.


## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

## 2. Load the Breast Cancer Dataset

```python
data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name="target")

print("Dataset shape:", X.shape)
print("Target shape:", y.shape)

X.head()
```

## 3. Check Missing Values

We check the dataset for missing values before training the models.

```python
print("Total missing values:", X.isnull().sum().sum())
print("\nMissing values in each column:")
print(X.isnull().sum())
```

### Preprocessing Explanation

- The dataset is checked for missing values.
- `SimpleImputer(strategy="median")` is included in the pipelines to handle any missing numerical values.
- `StandardScaler()` is used for Logistic Regression, SVM, and k-NN because these algorithms are affected by feature scale.
- Decision Tree and Random Forest do not require feature scaling.


## 4. Split the Dataset into Training and Testing Data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## 5. Logistic Regression

**How it works:** Logistic Regression predicts the probability of an observation belonging to a class.

**Why suitable:** The Breast Cancer dataset is a binary classification problem, so Logistic Regression is suitable.


```python
logistic_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=5000))
])

logistic_model.fit(X_train, y_train)
y_pred_logistic = logistic_model.predict(X_test)

print("Logistic Regression Accuracy:",
      accuracy_score(y_test, y_pred_logistic))
```

## 6. Decision Tree Classifier

**How it works:** A Decision Tree repeatedly splits the data based on feature values to make a prediction.

**Why suitable:** It can model nonlinear relationships and does not require feature scaling.


```python
decision_tree = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", DecisionTreeClassifier(random_state=42))
])

decision_tree.fit(X_train, y_train)
y_pred_tree = decision_tree.predict(X_test)

print("Decision Tree Accuracy:",
      accuracy_score(y_test, y_pred_tree))
```

## 7. Random Forest Classifier

**How it works:** Random Forest combines predictions from multiple decision trees.

**Why suitable:** It can handle many numerical features and capture complex relationships.


```python
random_forest = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ))
])

random_forest.fit(X_train, y_train)
y_pred_rf = random_forest.predict(X_test)

print("Random Forest Accuracy:",
      accuracy_score(y_test, y_pred_rf))
```

## 8. Support Vector Machine (SVM)

**How it works:** SVM finds a decision boundary that separates classes while maximizing the margin between them.

**Why suitable:** SVM can work effectively with numerical features and complex class boundaries. Feature scaling is important for SVM.


```python
svm_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", SVC())
])

svm_model.fit(X_train, y_train)
y_pred_svm = svm_model.predict(X_test)

print("SVM Accuracy:",
      accuracy_score(y_test, y_pred_svm))
```

## 9. k-Nearest Neighbors (k-NN)

**How it works:** k-NN classifies a new observation based on the classes of its nearest neighboring observations.

**Why suitable:** The dataset contains numerical features, and k-NN can use distances between observations. Feature scaling is important.


```python
knn_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier(n_neighbors=5))
])

knn_model.fit(X_train, y_train)
y_pred_knn = knn_model.predict(X_test)

print("k-NN Accuracy:",
      accuracy_score(y_test, y_pred_knn))
```

## 10. Model Comparison

```python
models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree,
    "Random Forest": random_forest,
    "SVM": svm_model,
    "k-NN": knn_model
}

results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred)
    })

results_df = pd.DataFrame(results)

results_df
```

## 11. Identify Highest and Lowest Accuracy

The following code identifies the models with the highest and lowest test accuracy for the selected train-test split.


```python
best_model = results_df.loc[results_df["Accuracy"].idxmax()]
worst_model = results_df.loc[results_df["Accuracy"].idxmin()]

print("Highest Accuracy:")
print(best_model)

print("\nLowest Accuracy:")
print(worst_model)
```

## 12. Conclusion

Five supervised classification algorithms were implemented on the Breast Cancer dataset: Logistic Regression, Decision Tree, Random Forest, SVM, and k-NN.

The dataset was checked for missing values and preprocessing was performed using median imputation and feature scaling where required. The models were evaluated using accuracy, precision, recall, and F1-score.

The final best and worst performers are determined directly from the test-set results above. Results can vary depending on the train-test split and model settings.
